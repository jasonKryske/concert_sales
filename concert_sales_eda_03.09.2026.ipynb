```python
%sql

-- Removed upsells as I found them just to explore the data a little bit more after seeing the how dirty the field is

SELECT DISTINCT event_name from events

WHERE 
  lower(event_name) NOT like '%not a concert ticket%' 
  AND 
  lower(event_name) NOT LIKE '%parking%'
  AND 
  lower(event_name) NOT LIKE '%extras%'
    AND 
  lower(event_name) NOT LIKE '%lockers%'
    AND
  lower(event_name) NOT like '%not an event%' 
   AND
  lower(event_name) NOT like '%coat check%'
   AND
  lower(event_name) NOT like '%upsells%'
   AND
  lower(event_name) NOT like '%vip%' 
   AND
  lower(event_name) NOT like '%access%' 
 AND
  lower(event_name) NOT like '%premium%'
 AND
  lower(event_name) NOT like '%suite%'
 AND
  lower(event_name) NOT like '%box seats%'
 AND
  lower(event_name) NOT like '%fast lane%'
   AND
  lower(event_name) NOT like '%commemorative ticket%'
```
